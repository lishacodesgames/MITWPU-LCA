```python
# Sum of all numbers in list

def sum_list(arr):
    sum = 0
    for elm in arr:
        sum += elm

    return sum

arr = map(int, input("Enter a few numbers: ").split())
print("Sum = ", sum_list(arr))
```

```python
# Product of all numbers in list

def product_list(arr):
    prod = 1
    for elm in arr:
        prod *= elm

    return prod

arr = map(int, input("Enter a few numbers: ").split())
print("Product = ", product_list(arr))
```

```python
# Reverse a string

def reverse_string(string):
    new_str = ""
    list_str = list(string)
    for char in list_str[::-1]:
        new_str += char

    return new_str

string = reverse_string("1234567890")
print(string)
```

```python
# Calculate factorial of number

def factorial(num):
    fact = 1
    for i in range(2, num + 1):
        fact *= i

    return fact

num = int(input("Enter a number: "))
print(f"{num}! = {factorial(num)}")
```

```python
# Check if a number is prime
import math

def is_prime(num):
    if num % 2 == 0 and num > 2:
        return False

    for i in range(3, int(math.sqrt(num)), 2):
        if num % i == 0:
            return False

    return True

num = int(input("Enter a number: "))
if num < 2:
    print("Number is neither prime nor composite")
elif is_prime(num):
    print("Number is prime")
else:
    print("Number is not prime")
```

```python
# Check if string is palindrome
def is_palindrome(string):
    _list = list(string)

    for i in range(len(string) // 2):
        if _list[i] != _list[-(i+1)]:
            return False

    return True

string = input("Enter a string: ")
if is_palindrome(string):
    print("It is a palindrome")
else:
    print("It is not a palindrome")
```

```python
# Print even numbers from a given list
def print_evens(list_nums):
    for num in list_nums:
        if num % 2 == 0:
            print(num, end=" ")

nums = map(int, input("Enter a few numbers: ").split())
print_evens(nums)
```
