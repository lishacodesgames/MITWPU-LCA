```python
# ---------- TUPLE OPERATIONS ----------

t = (10, 20, 30, 40, 20)

print("Original Tuple:", t)
print("Element at index 2:", t[2])
print("Count of 20:", t.count(20))
print("Index of 30:", t.index(30))

t2 = t + (50, 60)
print("After Concatenation:", t2)
print("Sliced Tuple (1:4):", t[1:4])
```

```python
# ---------- DICTIONARY OPERATIONS ----------

student = {"name": "Lisha", "age": 17, "branch": "CSE"}

print("Original Dictionary:", student)

student["college"] = "MIT-WPU"
print("After Adding 'college':", student)

student.update({"name": "Lisha Patil"})
print("After Updating 'name':", student)

print("Name:", student.get("name"))

student.pop("branch")
print("After Removing 'branch':", student)

print("Keys:", list(student.keys()))
print("Values:", list(student.values()))
print("Items:", list(student.items()))

print("'name' in dictionary?", "name" in student)

del student["college"]
print("After del 'college':", student)
```
