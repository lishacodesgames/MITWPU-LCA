```python
# check largest of 3 numbers

a, b, c = map(int, input("Enter three numbers: ").split())
largest_str = "%s is the largest"

if a > b:
    if a > c:
        print(largest_str % a)
    else:
        print(largest_str % c)
else:
    if b > c:
        print(largest_str % b)
    else:
        print(largest_str % c)
```
